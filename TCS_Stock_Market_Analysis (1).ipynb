3. Download Stock Data




---



```python

import yfinance as yf
import pandas as pd
import numpy as np
import plotly.express as px

# Download TCS stock data
stock = yf.download(
    "TCS.NS",
    period="1y",
    auto_adjust=True
)

# Simplify yfinance's multi-level column names
if isinstance(stock.columns, pd.MultiIndex):
    stock.columns = stock.columns.get_level_values(0)

# Remove rows with missing closing prices
stock = stock.dropna(subset=["Close"]).copy()

# Prepare data for charts and analysis
chart_data = stock.reset_index()

print(stock.tail())
print("\nMissing closing prices:", stock["Close"].isna().sum())
```

4. Visualize Stock Prices and Moving Averages


---



```python
print(chart_data[["Date", "Close"]].tail(10))
```

```python
stock["Moving_Average_20"] = (
    stock["Close"].rolling(window=20).mean()
)

print(stock[["Close", "Moving_Average_20"]].tail())
```

```python
import plotly.express as px

chart_data = stock.reset_index()

# Handle multi-level columns from yfinance
if isinstance(chart_data.columns, pd.MultiIndex):
    chart_data.columns = chart_data.columns.get_level_values(0)

fig = px.line(
    chart_data,
    x="Date",
    y="Close",
    title="TCS Historical Stock Price",
    labels={
        "Date": "Date",
        "Close": "Closing Price (₹)"
    }
)

fig.show()
```

```python
chart_data["Moving_Average_20"] = (
    chart_data["Close"].rolling(window=20).mean()
)

chart_data[["Date", "Close", "Moving_Average_20"]].tail()
```

```python
fig = px.line(
    chart_data,
    x="Date",
    y=["Close", "Moving_Average_20"],
    title="TCS Closing Price vs 20-Day Moving Average",
    labels={
        "Date": "Date",
        "value": "Price (₹)",
        "variable": "Metric"
    }
)

fig.show()
```

5. Closing Price Analysis


---





```python
first_price = chart_data["Close"].iloc[0]
last_price = chart_data["Close"].iloc[-1]

total_return = ((last_price - first_price) / first_price) * 100

print(f"First closing price: ₹{first_price:.2f}")
print(f"Last closing price: ₹{last_price:.2f}")
print(f"Percentage change: {total_return:.2f}%")
```

```python
highest_price = chart_data["Close"].max()
lowest_price = chart_data["Close"].min()

print(f"Highest closing price: ₹{highest_price:.2f}")
print(f"Lowest closing price: ₹{lowest_price:.2f}")
```

6. Daily Returns Analysis


---



```python
close = stock["Close"]

# Calculate daily price change and percentage return
chart_data["Daily_Change"] = chart_data["Close"].diff()

chart_data["Daily_Return_%"] = (
    chart_data["Close"].pct_change(fill_method=None) * 100
)

print(
    chart_data[
        ["Date", "Close", "Daily_Change", "Daily_Return_%"]
    ].tail()
)
```

```python
daily_returns = chart_data["Daily_Return_%"].dropna()

average_return = np.mean(daily_returns)

print(f"Average daily return: {average_return:.4f}%")
```

7. Volatility Analysis


---



```python
volatility = np.std(daily_returns)

print(f"Daily return volatility: {volatility:.4f}%")
```

8. Final Summary and Conclusion


---



```python

import numpy as np
import pandas as pd

# Remove rows where closing price is missing
chart_data = chart_data.dropna(subset=["Close"]).copy()

# Calculate daily returns if not already calculated
chart_data["Daily_Return_%"] = (
    chart_data["Close"].pct_change() * 100
)

# Calculate key statistics
first_price = chart_data["Close"].iloc[0]
last_price = chart_data["Close"].iloc[-1]

percentage_change = (
    (last_price - first_price) / first_price
) * 100

highest_price = chart_data["Close"].max()
lowest_price = chart_data["Close"].min()

daily_returns = chart_data["Daily_Return_%"].dropna()

average_daily_return = np.mean(daily_returns)
volatility = np.std(daily_returns)

# Display the automatic summary
print("===== TCS STOCK ANALYSIS SUMMARY =====\n")

print(f"First closing price: ₹{first_price:.2f}")
print(f"Latest closing price: ₹{last_price:.2f}")
print(f"Overall price change: {percentage_change:.2f}%")
print(f"Highest closing price: ₹{highest_price:.2f}")
print(f"Lowest closing price: ₹{lowest_price:.2f}")
print(f"Average daily return: {average_daily_return:.4f}%")
print(f"Daily return volatility: {volatility:.4f}%")

print("\n===== INTERPRETATION =====")

if percentage_change > 0:
    print("The stock price increased over the selected period.")
elif percentage_change < 0:
    print("The stock price decreased over the selected period.")
else:
    print("The stock price remained unchanged over the selected period.")

if volatility > 2:
    print("Daily returns fluctuated considerably during this period.")
else:
    print("Daily returns showed relatively lower variability during this period.")
```
